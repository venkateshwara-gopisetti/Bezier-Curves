# beziervis quickstart

Evaluate Bézier curves with NumPy and watch de Casteljau's algorithm build them.

```python
%pip install -q "beziervis[all]"
```

```python
import numpy as np
import beziervis as bv

ctrl = [[0, 0], [5, 10], [9, 7], [4, 3]]
pts = bv.points(ctrl, n=200)
pts.shape, pts[:3]
```

## Watch it build

The last expression in a cell plays inline. `verbose=True` draws every level of the construction.

```python
bv.animate(ctrl, verbose=True)
```

## The construction at one value of t

Level 0 is the control polygon. Each later level interpolates between neighbouring points of the level before, and the final level is a single point on the curve.

```python
for level, p in enumerate(bv.casteljau_steps(ctrl, t=0.4)):
    print(level, p.round(3).tolist())
```

## Operations

```python
left, right = bv.split(ctrl, 0.5)
print('left :', left.tolist())
print('right:', right.tolist())
print('derivative control points:', bv.derivative(ctrl).tolist())
print('arc length:', bv.arc_length(ctrl))
```

## Any dimension, many curves

```python
space_curve = np.random.default_rng(0).normal(size=(6, 3))
print(bv.points(space_curve, n=5).shape)

curves = np.random.default_rng(1).normal(size=(10_000, 4, 2))
print(bv.points_batch(curves, n=100).shape, 'using', bv.get_backend())
```
